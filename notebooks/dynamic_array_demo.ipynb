```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "dynamic_array" / "__init__.py").is_file():
    project_root = project_root.parent
if not (project_root / "dynamic_array" / "__init__.py").is_file():
    raise FileNotFoundError("Could not locate the project's dynamic_array package.")
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from dynamic_array import MyList

values = MyList()
for item in (10, 20, 30):
    values.append(item)

print(values)
print(f"Length: {len(values)}, capacity: {values.size}")
```
