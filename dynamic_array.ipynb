```python
import ctypes
```

```python
# it create dynamic array
class MyList:
  def __init__(self):
    self.size = 1
    self.n = 0
    # create a c type array with size = self.size
    self.A = self.__make_array(self.size)


  def __str__(self):
    # list like that [1,2,3,4]
    result = ''
    for i in range(self.n):
      result += str(self.A[i]) + ','

    return '[' +result[:-1] + ']'


  def __getitem__(self,index):
    if 0 <= index < self.n:
      return self.A[index]
    else:
      return 'IndexError - Index out of range'


  # ------- len() -----------
  def __len__(self):
    return self.n


  # ------ append() --------
  def append(self,item):
    if self.n == self.size:
      #resize
      self.__resize(self.size*2)

    # append
    self.A[self.n] = item
    self.n += 1


  # ------- pop()-------
  def pop(self):
    if self.n == 0:
      return 'Empty list'
    print(self.A[self.n-1])
    self.n -= 1


  # ------- clear() --------
  def clear(self):
    self.n = 0
    self.size = 1


  # ------ find() -----------
  def find(self,item):
    for i in range(self.n):
      if self.A[i] == item:
        return i
    return 'ValueError - item is not in the list'


  # ------ insert() -----------
  def insert(self,pos,item):
    if self.n == self.size:
      self.__resize(self.size*2)

    for i in range(self.n,pos,-1):
      self.A[i] = self.A[i-1]

    self.A[pos] = item
    self.n += 1


  # ----- del() -----------------
  def __delitem__(self,pos):
    if 0<= pos <self.n:
      for i in range(pos,self.n-1):
        self.A[i] = self.A[i+1]

      self.n -= 1
    else:
      return 'IndexError - Index out of range'


  # ------------ remove() ------------
  def remove(self,item):
    pos = self.find(item)
    if type(pos) == int:
      # delete
      self.__delitem__(pos)
    else:
      return pos

  # ------- min() -------
  def min(self):
    m = self.A[0]
    for i in range(self.n):
      if self.A[i] < m:
        m = self.A[i]
    return m

  # ------- max() ---------
  def max(self):
    m = self.A[0]
    for i in range(self.n):
      if self.A[i] > m:
        m = self.A[i]
    return m

  # -------- sum() ----------
  def sum(self):
    s = 0
    for i in range(self.n):
      s += self.A[i]
    return s

  # ------ sort() -------
  def sort(self):
    for i in range(self.n-1):
      for j in range(i+1,self.n):
        if self.A[i] > self.A[j]:
          self.A[i],self.A[j] = self.A[j],self.A[i]

  # ------- merge() --------
  def merge(self,list2):
    # We need to ensure, we have enough space to store
    total_elements = self.n + list2.n
    if self.size < total_elements:
      self.__resize(total_elements * 2) # Double the needed capacity for future growth

    for i in range(list2.n):
      self.A[self.n] = list2.A[i]
      self.n += 1

  # ------ extend() ---------
  def extend(self,list2):
    self.merge(list2)




  def __resize(self,new_capacity):
    # create a new array with new capacity
    B = self.__make_array(new_capacity)
    self.size = new_capacity

    # copy the content of A to B
    for i in range(self.n):
      B[i] = self.A[i]

    # reassign A
    self.A = B


  def __make_array(self, captacity):
    return(captacity*ctypes.py_object)()
```
